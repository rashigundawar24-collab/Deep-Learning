#Assignmenet 1 - Install and configure TensorFlow/Keras in Google Colab. Perform data preprocessing, normalization, train-test split, and visualization on a sample dataset.

```python
import tensorflow as tf
import matplotlib.pyplot as plt
```

```python
# Get data and split into train and test datasets
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.mnist.load_data()
```

```python
# Scale down values of pixels from 0-255 to 0-1
train_images = train_images / 255.0
test_images = test_images / 255.0
```

```python
# Visualizing data
print("Training data shape:", train_images.shape)
print("Testing data shape:", test_images.shape)
print("Training labels:", train_labels)

plt.imshow(train_images[5], cmap='gray')
plt.title("Sample Digit")
plt.show()
```

```python
# Defining neural network model
my_model = tf.keras.models.Sequential()

my_model.add(tf.keras.layers.Flatten(input_shape=(28, 28)))
my_model.add(tf.keras.layers.Dense(100, activation='relu'))
my_model.add(tf.keras.layers.Dense(10, activation='softmax'))
```

```python
# Compiling the model
my_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
# Training the model
my_model.fit(train_images, train_labels, epochs=4)
```

```python
# Checking model accuracy on test data
test_loss, test_accuracy = my_model.evaluate(test_images, test_labels)

print("Test accuracy:", test_accuracy)
```

```python
# Saving the model
my_model.save('my_mnist_model.keras')
```
